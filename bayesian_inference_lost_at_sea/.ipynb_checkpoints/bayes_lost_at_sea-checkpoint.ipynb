```python
from math import sqrt
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
import pandas as pd
import random
import seaborn as sns
```

# Simulating the drift of the individual on the SUP

```python
'''
last-known-position lkp: Tuple of (lat, lon)

current_velocity: Tuple (eastward m/s, northward m/s)

dt: time step in seconds (e.g., 60s)

steps: e.g., 120 steps for 2 hours @ 1 min intervals
'''

def simulate_drift(lkp, current_velocity, dt, steps, noise_std=0.5):
    path = [lkp]
    lat, lon = lkp
    for _ in range(steps):
        # Approximate drift in meters, convert to degrees
        drift_east = current_velocity[0] * dt + np.random.normal(0, noise_std)
        drift_north = current_velocity[1] * dt + np.random.normal(0, noise_std)

        lat += drift_north / 111_000  # degrees per meter latitude
        lon += drift_east / (111_000 * np.cos(np.radians(lat)))  # degrees per meter longitude

        path.append((lat, lon))
    return pd.DataFrame(path, columns=["latitude", "longitude"])
```

### Create the prior for the entire grid

```python
def create_prior(grid_size=1000, resolution=25, std_dev=150):
    grid_N = int(grid_size / resolution)
    x = np.linspace(-grid_size/2, grid_size/2, grid_N)
    y = np.linspace(-grid_size/2, grid_size/2, grid_N)
    xv, yv = np.meshgrid(x, y)

    # Gaussian prior centered at (0, 0)
    prior = np.exp(-(xv**2 + yv**2) / (2 * std_dev**2))
    prior /= prior.sum()  # Normalize

    return prior, x, y

prior, x, y = create_prior()
```

```python
def update_posterior(posterior_grid, searched_cells, detection_prob=0.9):
    updated_grid = posterior_grid.copy()
    for x, y in searched_cells:
        updated_grid[x, y] *= (1 - detection_prob)
    
    # Normalize to make it a valid probability distribution
    updated_grid /= updated_grid.sum()
    return updated_grid
```

### Search grid

```python
grid_size = 1000        # meters (1 km x 1 km area)
resolution = 25         # meters per grid cell
n_cells = grid_size // resolution
search_grid = []
for x in range(n_cells):
    for y in range(n_cells):
        search_grid.append((x, y))
search_grid = np.array(search_grid)
```

```python
def get_top_k_cells(prior_grid, k=10):
    flat_indices = np.argsort(prior_grid.ravel())[::-1]
    top_k = np.column_stack(np.unravel_index(flat_indices[:k], prior_grid.shape))
    return [tuple(cell) for cell in top_k]
```

```python
def get_top_k_cells_with_radius(posterior_grid, k=10, radius=1):
    # Flatten and sort the indices
    flat_indices = np.dstack(np.unravel_index(np.argsort(posterior_grid.ravel())[::-1], posterior_grid.shape))[0]

    selected_cells = []
    seen_cells = set()

    for center in flat_indices:
        x_c, y_c = center
        local_cells = []

        for dx in range(-radius, radius + 1):
            for dy in range(-radius, radius + 1):
                x, y = x_c + dx, y_c + dy
                if 0 <= x < posterior_grid.shape[0] and 0 <= y < posterior_grid.shape[1]:
                    if (x, y) not in seen_cells:
                        local_cells.append((x, y))
                        seen_cells.add((x, y))

        selected_cells.extend(local_cells)
        if len(seen_cells) >= k * (2 * radius + 1) ** 2:
            break

    return selected_cells
```

```python
# Parameters
grid_size_m = 1000
resolution_m = 25
grid_cells = int(grid_size_m / resolution_m)  # 40x40 grid
n_paths = 1000
n_steps = 480  # Simulating 480 minutes, 1 step per minute
dt = 60  # seconds per step

# Start position
start_pos = np.array([grid_cells / 20, 4 * grid_cells / 23])  # ~ (2, 7)

# Drift parameters
drift_velocity = np.array([0.05, 0.01])  # cells per step (eastward + slight north)
noise_std = 0.1  # randomness in movement

# Generate 1000 simulated paths
all_endpoints = []
for _ in range(n_paths):
    pos = start_pos.copy()
    for _ in range(n_steps):
        noise = np.random.normal(0, noise_std, size=2)
        pos += drift_velocity + noise
        # Clip to stay within bounds
        pos = np.clip(pos, 0, grid_cells - 1)
    end_cell = tuple(np.round(pos).astype(int))
    all_endpoints.append(end_cell)

# Convert to DataFrame for counting
df = pd.DataFrame(all_endpoints, columns=['x', 'y'])
prior_counts = df.value_counts().reset_index(name='count')

# Build prior probability grid
prior_grid = np.zeros((grid_cells, grid_cells))
for _, row in prior_counts.iterrows():
    prior_grid[row['y'], row['x']] = row['count']
prior_grid = prior_grid / np.sum(prior_grid)

# Generate the actual path (for illustration)
actual_path = [start_pos.copy()]
pos = start_pos.copy()
for _ in range(n_steps):
    noise = np.random.normal(0, noise_std, size=2)
    pos += drift_velocity + noise
    pos = np.clip(pos, 0, grid_cells - 1)
    actual_path.append(pos.copy())
actual_path = np.array(actual_path)
end_pos =  actual_path[-1]

# Convert to DataFrame for counting
df = pd.DataFrame(all_endpoints, columns=['x', 'y'])
prior_counts = df.value_counts().reset_index(name='count')

# Build prior probability grid
prior_grid = np.zeros((n_cells, n_cells))
for _, row in prior_counts.iterrows():
    prior_grid[row['y'], row['x']] = row['count']
prior_grid = prior_grid / np.sum(prior_grid)


plt.figure(figsize=(8, 6))
sns.heatmap(prior_grid, cmap='YlOrRd', cbar_kws={'label': 'Prior Probability'})
plt.plot(actual_path[:, 0], actual_path[:, 1], color='blue', label='Actual Drift Path', linewidth=2)
plt.scatter(*start_pos, color='green', label='Start Pos', zorder=5)
plt.scatter(*end_pos, color='yellow', label='End Pos', zorder=5)
plt.legend()
plt.title("Prior Probability Heatmap from 1000 Drift Paths")
plt.xlabel("East (Grid X)")
plt.ylabel("North (Grid Y)")
plt.gca().invert_yaxis()
plt.show()
```

```python
def apply_search_penalty(posterior_grid, zone_centers, radius, penalty=0.1):
    for x, y in zone_centers:
        x_min = max(x - radius, 0)
        x_max = min(x + radius + 1, posterior_grid.shape[1])
        y_min = max(y - radius, 0)
        y_max = min(y + radius + 1, posterior_grid.shape[0])
        posterior_grid[y_min:y_max, x_min:x_max] *= penalty
    posterior_grid /= posterior_grid.sum()  # renormalize
    return posterior_grid
```

```python
# Reinitialize posterior from prior
posterior_grid = prior_grid.copy()

# Keep track of all searched cells
all_searched_cells = []
all_zone_centers = []
# Run 5 sequential search rounds
for round_num in range(30):
    print(f"Search round {round_num + 1}")

    # Select top K cells to search
    searched_cells = get_top_k_cells_with_radius(posterior_grid, k=2, radius=1)  # 2 zones of 3x3
    all_searched_cells.extend(searched_cells)

     # Select top zones to search
    zone_centers = get_top_k_cells(posterior_grid, k=2)
    all_zone_centers.extend(zone_centers)
    all_zone_centers = list(set(all_zone_centers))
    searched_cells = get_top_k_cells_with_radius(posterior_grid, k=2, radius=1)
    
    # Apply penalty for revisiting
    posterior_grid = apply_search_penalty(posterior_grid, zone_centers, radius=1, penalty=0.05)
    # Update posterior based on these search attempts
    posterior_grid = update_posterior(posterior_grid, searched_cells, detection_prob=0.9)

    # Visualize updated posterior
    plt.figure(figsize=(8, 6))
   
    
    plt.title(f'Posterior After Search Round {round_num + 1}')
    plt.xlabel('East (Grid X)')
    plt.ylabel('North (Grid Y)')
    plt.scatter(*start_pos, color='green', label='Start Pos', zorder=5)
    plt.plot(actual_path[:, 0], actual_path[:, 1], color='blue', label='Actual Drift Path', linewidth=2)
    plt.scatter(*end_pos, color='yellow', label='End Pos', zorder=5)

   

    # Check for detection
    # Ensure end_pos is a tuple of ints
    end_pos[0] = int(round(end_pos[0]))
    end_pos[1] = int(round(end_pos[1]))
    print(end_pos)
    end_cell = tuple(map(int, [end_pos[0],end_pos[1]]))
    
    
    # Ensure searched_cells are also tuples of ints
    searched_cells = [tuple(map(int, [cell[1],cell[0]])) for cell in searched_cells]

    near_miss_thresh = 1
    
    if end_cell in searched_cells:
        found = random.random() < 0.9  # Simulate detection with 0.9 probability
        if found:
            print(f"\n🎯 Target found at {end_pos} in round {round_num + 1}!")
            posterior_grid[int(end_cell[1]), int(end_cell[0])] = 1.0  # certainty
            sns.heatmap(posterior_grid, cmap='YlGnBu', square=True, cbar_kws={'label': 'Posterior Probability'})
            # Draw all past search zones as faint grey
            
            for x, y in all_zone_centers:
                rect = patches.Rectangle(
                    (y - 1 + 0.5, x - 1 + 0.5), 3, 3,
                    linewidth=1,
                    edgecolor='lightgrey',
                    facecolor='none'
                )
                plt.gca().add_patch(rect)
                
            # Draw red boxes around searched cells
            radius = 1
            for x, y in zone_centers:
                rect = patches.Rectangle(
                    (y - radius + 0.5, x - radius + 0.5),  # center shifted by radius and aligned to cell centers
                    width=2 * radius + 1,
                    height=2 * radius + 1,
                    linewidth=1.5,
                    edgecolor='red',
                    facecolor='none'
                )
                plt.gca().add_patch(rect)
            
            plt.gca().invert_yaxis()
            plt.legend()
            #plt.xlim(14, 36)       # x-axis (East direction)
            #plt.ylim(5, 24)       # y-axis (North direction, inverted)
            plt.show()
            break
        else:
            print(f"🔍 Target was in search area but not detected in round {round_num + 1}.")
    else:
        min_dist = 1000
        closest_cells = []
        for cell in searched_cells:
            if abs(cell[0] - end_cell[0]) <= near_miss_thresh and abs(cell[1] - end_cell[1]) <= near_miss_thresh:
                print(f"👀 Near miss: {cell} close to {end_cell}")
        dist = sqrt((cell[0] - end_cell[0])**2 + (cell[1] - end_cell[1])**2)
        if dist < min_dist:
            closest_cells = [cell, end_cell]
            dist = min_dist
        
        #print(f"min_dist = {min_dist:.2f}")
        print(closest_cells)
    

        
    sns.heatmap(posterior_grid, cmap='YlGnBu', square=True, cbar_kws={'label': 'Posterior Probability'})
    # Draw all past search zones as faint grey
    for x, y in all_zone_centers:
        rect = patches.Rectangle(
            (y - 1 + 0.5, x - 1 + 0.5), 3, 3,
            linewidth=1,
            edgecolor='lightgrey',
            facecolor='none'
        )
        plt.gca().add_patch(rect)
        
    # Draw red boxes around searched cells
    radius = 1
    for x, y in zone_centers:
        rect = patches.Rectangle(
            (y - radius + 0.5, x - radius + 0.5),  # center shifted by radius and aligned to cell centers
            width=2 * radius + 1,
            height=2 * radius + 1,
            linewidth=1.5,
            edgecolor='red',
            facecolor='none'
        )
        plt.gca().add_patch(rect)
    plt.gca().invert_yaxis()
    plt.legend()
    #plt.xlim(14, 36)       # x-axis (East direction)
    #plt.ylim(5, 24)       # y-axis (North direction, inverted)
    plt.show()
```

```python
def plot_heatmap(prior, x, y, title='Prior Probability Heatmap'):
    plt.figure(figsize=(6, 5))
    plt.contourf(x, y, prior, levels=50, cmap='hot')
    plt.colorbar(label='Probability')
    plt.title(title)
    plt.xlabel('East-West Offset (m)')
    plt.ylabel('North-South Offset (m)')
    plt.axis('equal')
    plt.grid(True)
    plt.show()

plot_heatmap(prior, x, y)
```
